```python
import numpy as np
from helper import sigmoid, grade_hidden_layer, grade_output_layer

# Network size
N_input  = 4
N_hidden = 3
N_output = 2

np.random.seed(42)

# Fake input data and weights
X = np.random.randn(4)
weights_input_to_hidden  = np.random.normal(0, scale=0.1, size=(N_input, N_hidden))
weights_hidden_to_output = np.random.normal(0, scale=0.1, size=(N_hidden, N_output))
```

```python
# TODO: Make a forward pass through the hidden layer
#       hidden_layer_in  = dot product of X and weights_input_to_hidden
#       hidden_layer_out = sigmoid of hidden_layer_in
hidden_layer_in  = np.dot(X, weights_input_to_hidden)  # replace with your code
hidden_layer_out = sigmoid(hidden_layer_in)  # replace with your code

print('Hidden-layer Output:')
print(hidden_layer_out)

# ── Grading ──────────────────────────────────────────────────────────────────
grade_hidden_layer(hidden_layer_in, hidden_layer_out, X, weights_input_to_hidden)
```

```python
# TODO: Make a forward pass through the output layer
#       output_layer_in  = dot product of hidden_layer_out and weights_hidden_to_output
#       output_layer_out = sigmoid of output_layer_in
output_layer_in  = np.dot(hidden_layer_out, weights_hidden_to_output)  # replace with your code
output_layer_out = sigmoid(output_layer_in)  # replace with your code

print('Output-layer Output:')
print(output_layer_out)

# ── Grading ──────────────────────────────────────────────────────────────────
grade_output_layer(output_layer_in, output_layer_out,
                   hidden_layer_out, weights_hidden_to_output)
```
